# Content-Based Recommender
Feature engineering, TF-IDF, similarity demo, and save step.

```python
from pathlib import Path
from recommender.content_based import ContentBasedRecommender
from recommender.utils import load_data
p,u,i = load_data()
m = ContentBasedRecommender().fit(p)
display(p[['product_id','name','category','tags']].head())
print('vocabulary size', len(m.vectorizer.vocabulary_))
display(m.recommend('P0001', 10))
cold = p.loc[p.is_new_arrival, 'product_id'].iloc[0]
print('cold-start seed', cold, m.recommend(cold, 3))
Path('models').mkdir(exist_ok=True)
m.save(Path('models/content_based.pkl'))
```
