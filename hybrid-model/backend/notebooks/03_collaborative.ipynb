# Collaborative Recommender
Matrix construction, mean-centering, SVD, fallback, and save step.

```python
from pathlib import Path
from recommender.collaborative import CollaborativeRecommender
from recommender.utils import load_data
p,u,i = load_data()
m = CollaborativeRecommender().fit(i,p)
print('matrix', m.user_item_df.shape, 'predictions', m.predictions_df.shape)
display(m.predictions_df.iloc[:20,:30])
for user in ['U0001','U0042','U0100']:
    display(m.recommend(user, 5, interactions_df=i))
display(m.recommend('UNKNOWN', 5))
Path('models').mkdir(exist_ok=True)
m.save(Path('models/collaborative.pkl'))
```
