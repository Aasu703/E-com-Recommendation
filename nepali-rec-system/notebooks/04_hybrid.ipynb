# Hybrid Recommender
Adaptive alpha, festival mode, new arrivals, and save step.

```python
from pathlib import Path
from recommender.hybrid import HybridRecommender
from recommender.utils import load_data
p,u,i = load_data()
m = HybridRecommender().fit(p,u,i)
for user in ['U0001','U0042','U0200']:
    display(m.recommend(user, 5))
print('July')
display(m.recommend('U0042', 10, {'month':7}))
print('Dashain')
display(m.recommend('U0042', 10, {'month':10}))
print('new arrivals surfaced', sum(r['is_new_arrival'] for r in m.recommend('U0042', 50)))
Path('models').mkdir(exist_ok=True)
m.save(Path('models/hybrid_recommender.pkl'))
```
